```python
def initialLoop():
    print("#######################")
    print("#######################")
    print("#######################")
    print('Enter "exit" at any time to leave')
    print("Hello, welcome to my slot machine")
    print("First, input your name!")

    
    while(True):
        try:
            name = str(input("Name: "))
            break
        except ValueError:
            print("Enter a valid name or exit")
    if name == "exit":
        sys.exit("Goodbye!")
    print(f"{name.title()}, how much would you like to make for an initial deposit?")

    
    while(True):
        try:
            balance = input("Enter deposit: ")
            if balance == "exit":
                sys.exit("Goodbye!")
            balance = int(balance)
            if balance > 0:
                break
            else:
                print("Enter a number greater than 0")
        except ValueError:
            print("Enter an integer, or 'exit' to leave")
    return name, balance

    

def mainLoop(name, balance):
    
    while(True):
        print(f"Current balance: {balance}")
        
        while (balance == 0):
            print("You have no balance, would you like to deposit? Enter 'exit' to leave.")
            try:
                balance = input("Deposit: ")
                if balance == "exit":
                    sys.exit("Goodbye!")
                balance = int(balance)
                if balance <= 0:
                    print("Enter a number greater than 0")
                    balance = 0
            except ValueError:
                print("Enter a valid number, or 'exit' to leave")
                balance = 0
                
        try:
            print("How much would you like to bet?")
            bet = input("Bet: ")
            if bet == "exit":
                sys.exit(f"Your balance is {balance}. Goodbye {name.title()}")
            bet = int(bet)
            if bet < 0 or bet > balance:
                print("Enter a value greater than 0 or your balance")
                continue
                
        except ValueError:
            print("Enter valid number")
            continue

        balance -= bet

        result = Spin()
        final = checker(result)

        print(f"You won {final[1] * bet}!")
        balance += final[1] * bet
        
        
        
        
```
