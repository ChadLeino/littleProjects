```python
def analyze_voltage_readings(readings, lower_limit, upper_limit):
    final = {}
    final["passed"] = True
    final["failed_indices"] = []
    noneCount = 0
    if not readings:
        final["passed"] = False
        final["average_voltage"] = None
        return final
    for index, val in enumerate(readings):
        if val == None:
            final["failed_indices"].append(index)
            final["passed"] = False
            noneCount += 1
        else:
            val = val * 5
            final.setdefault("average_voltage", 0)
            final["average_voltage"] += val
            if val < lower_limit or val > upper_limit:
                final["passed"] = False
                final["failed_indices"].append(index)
    calc = final.setdefault("average_voltage")
    valid_count = len(readings) - noneCount

    if valid_count > 0:
        final["average_voltage"] /= valid_count
    else:
        final["average_count"] = None
    return final
    
```

```python
def find_trip_index(readings, lower_limit, upper_limit):
    consecutive = 0
    if not readings:
        raise ValueError("No readings")
    for index, value in enumerate(readings):
        if value == None:
            consecutive += 1
        elif value > upper_limit or value < lower_limit:
            consecutive += 1
        else:
            consecutive = 0
        if consecutive == 3:
            return index
    return None
```

```python
def find_trip_index(readings, lower_limit, upper_limit, trip_count):
    if trip_count <= 0:
        raise ValueError("trip_count must be positive")
    consecutive = 0
    if not readings:
        raise ValueError("No readings")
    for index, value in enumerate(readings):
        if value == None:
            consecutive += 1
        elif value > upper_limit or value < lower_limit:
            consecutive += 1
        else:
            consecutive = 0
        if consecutive == trip_count:
            return index
    return None   
```
