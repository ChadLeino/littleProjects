```python
class DewPointSensor:
    def __init__(self, name, minimumSafeDewpoint):
        self.name = name
        self.minimumSafeDewpoint = minimumSafeDewpoint
        self.readings = []

    def addReading(self, reading):
        self.readings.append(reading)

    def getAverage(self):
        return sum(self.readings) / len(self.readings)

    def isSafe(self):
        if not self.readings:
            return False
        return (self.minimumSafeDewpoint <= self.readings[-1])

    def displayStatus(self):
        print(self.name, "is", self.isSafe())

    
```

```python
x = DewPointSensor('s1', 0)

print(x.isSafe())
for j in range(0, 100, 10):
    x.addReading(j)


x.readings
```
