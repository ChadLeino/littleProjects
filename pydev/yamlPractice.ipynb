```python
import yaml
```

```python
with open("config.yaml", "r", encoding="utf-8") as file:
    config = yaml.safe_load(file)
```

```python
config.values()
```

```python
config['gui']['title']
```

```python
if config['tests']['power_supply']['enabled']:
    voltage = config['tests']['power_supply']['voltage']
    print(f"Running power at {voltage} V")
else:
    print("Power supply disabled")
    voltage = 0
```

```python
for test_name, settings in config.items():
    print(test_name)
    print(settings)
```

```python
for test_name, settings in config["tests"].items():
    if settings["enabled"]:
        print(f"RUN {test_name}")
    else:
        print(f"SKIP {test_name}")
```
